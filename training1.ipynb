Связь с numpy

```python
import numpy as np
```

```python
# конвертация torch -> numpy
a = torch.ones(5)
print(a)
b = a.numpy()
print(b)
print('\n')
# как меняется массив при изменении тензора (по ссылке)
a += 1
print(a)
print(b)
```

```python
# обратная конвертация
a = np.ones(5)
b = torch.from_numpy(a)
print(a)
print(b)
print('\n')
# как меняется тензор при изменении массива (по ссылке)
a += 1
print(a)
print(b)
```

Конкатенация тензоров

```python
# конкатенация тензоров вдоль оси 0
first_1 = torch.randn(2, 5)
print(first_1)
second_1 = torch.randn(3, 5)
print(second_1)
concat_1 = torch.cat([first_1, second_1], 0) # dim=0 default
print(concat_1)
print(concat_1.shape)
print('\n')
# конкатенация тензоров вдоль оси 1
first_2 = torch.randn(2, 3)
print(first_2)
second_2 = torch.randn(2, 5)
print(second_2)
concat_2 = torch.cat([first_2, second_2], 1)
print(concat_2)
print(concat_2.shape)
```

```python
# добавляем измерение в тензор
tensor_1 = torch.tensor([1, 2, 3, 4])
tensor_a = torch.unsqueeze(tensor_1, 0) # на какую позицию добавляется новое измерение ([4] -> [1, 4])
print(tensor_a)
print(tensor_a.shape)
print('\n')
tensor_b = torch.unsqueeze(tensor_1, 1)
print(tensor_b)
print(tensor_b.shape)
print('\n')
tensor_2 = torch.rand(2, 3, 4)
tensor_c = tensor_2[:, :, 2]
print(tensor_c)
print(tensor_c.shape)
print('\n')
tensor_d = torch.unsqueeze(tensor_c, 2)
print(tensor_d)
print(tensor_d.shape)
```

```python
# можно использовать метод
print(tensor_1.unsqueeze(1))
```

```python
# После require_grad=True тензор запоминает, как он создаётся и используется
x = torch.tensor([1., 2., 3.], requires_grad=True)
y = torch.tensor([4., 5., 6.], requires_grad=True)
z = x + y
print(z)
print(z.grad_fn)
print('\n')
# Проверяем другую функцию тензора
s = z.sum()
print(s)
print(s.grad_fn)
```

```python
# Считаем backprop (производная s - функция потерь по x)
s.backward()
print(x.grad)
```

```python
# requieres_grad по умолчанию false
x = torch.randn(2, 2)
y = torch.randn(2, 2)
print(x.requires_grad, y.requires_grad)
z = x + y
print(z.grad_fn)
print('\n')
# можно изменить этот параметр
x.requires_grad_()
y.requires_grad_()
z = x + y
print(z.grad_fn)
print(z.requires_grad)
print('\n')
# detach возвращает копию, но без истории вычисления
new_z = z.detach()
print(new_z.grad_fn)
print('\n')
# градиент можно временно отключать
print(x.requires_grad)
print((x + 10).requires_grad)
with torch.no_grad():
    print((x + 10).requires_grad)
```

```python
# пример
x = torch.ones(2, 2, requires_grad=True)
print(x)
y = x + 2
print(y)
print(y.grad_fn)
z = y * y * 3
out = z.mean()
print(z, out)
out.backward()
print(x.grad)
```
