```python
import torch
import torch.nn.functional as F
```

```python
# Outputs
predictions = torch.log(torch.rand(3))
predictions
```

```python
# Softmax
F.softmax(predictions, dim=-1)
```

```python
# Low temperature
F.softmax(predictions / 0.5, dim=-1)
```

```python
# High temperature
F.softmax(predictions / 5, dim=-1)
```

```python
# Almost uniform distribution
F.softmax(predictions / 100, dim=-1)
```
