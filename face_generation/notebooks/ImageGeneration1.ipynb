```python
import matplotlib.pyplot as plt
import torch
import torchvision
import torch_directml
from src.Generator import Generator
```

```python
device = torch.device('cpu')
if torch.cuda.is_available():
    device = torch.device('cuda')
    print('cuda')
elif torch_directml.is_available():
    device = torch_directml.device(0)
    print(torch_directml.device_name(0))
else:
    print('cpu')
```

```python
model = Generator()
checkpoint = torch.load('../models/model0.pth')
model.load_state_dict(checkpoint['generator'])
model = model.to(device)
```

```python
model.eval()
with torch.no_grad():
    z = torch.randn(16, 100, device=device)
    flags = torch.randint(0, 2, (16, 7), device=device).float()
    fake_images = model(z, flags)
    torchvision.utils.save_image(fake_images.cpu(),
                                 f'../generated_images/test.png',
                                 nrow=4,
                                 normalize=True)
```
