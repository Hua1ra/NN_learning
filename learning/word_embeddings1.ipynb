```python
import torch
import torch.nn as nn
import torch.nn.functional as f
import torch.optim as optim
```

```python
# The indices of the words represent the values in the lookup table
word_to_ind = {'I': 0, 'love': 1, 'eating': 2, 'and': 3, 'sleeping': 4}
embeddings = nn.Embedding(5, 7)
word_index = torch.tensor([word_to_ind['love']])
print(word_index)
```

```python
# Get the embeddibg
love = embeddings(word_index)
print(love)
print(love.shape)
```

```python
# Get all embeddings
all_ind = torch.tensor([w for w in range(5)], dtype=torch.long)
all_words = embeddings(all_ind)
print(all_ind)
print(all_words)
print(all_words.shape)
```
