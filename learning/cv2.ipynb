```python
import cv2 # type: ignore
```

```python
img = cv2.imread('source/images/car.jpg') # прочитать изображение (BGR)
```

```python
cv2.imshow('Image', img) # показать изображение
cv2.waitKey(0) # ждем нажатия любой клавиши
# можно указать кол-во мс для отображения
cv2.destroyAllWindows() # закрываем все окна
```

```python
cv2.imwrite('source/images/car_copy.jpg', img) # сохранить изображение
```

```python
converted_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB) # переводим к RGB
converted_img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # переводим к gray
_, bw_img = cv2.threshold(converted_img_gray, 127, 255, cv2.THRESH_BINARY) # Бинаризируем к black white
```

```python
cv2.imshow('Image', img)
cv2.waitKey(0)
cv2.imshow('Image', converted_img_gray)
cv2.waitKey(0)
cv2.imshow('Image', bw_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
resized_img = cv2.resize(img, (500, 500)) # изменение размера
cropped_img = img[50:100, 50:100]
```

```python
cv2.imshow('Image', resized_img)
cv2.waitKey(0)
cv2.imshow('Image', cropped_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Поворот
rows, cols = img.shape[0:2]
matrix = cv2.getRotationMatrix2D((cols / 2, rows / 2),
                                  45,
                                  1)
rotated_img = cv2.warpAffine(img, matrix, (cols, rows))
```

```python
cv2.imshow('Image', rotated_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
blurred_img = cv2.GaussianBlur(img, (5, 5), 0) # Блюр
```

```python
cv2.imshow('Image', blurred_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
draw = img.copy() # Копия для рисования
draw = cv2.line(draw, (0, 0), (200, 200), (255, 255, 255), 5)
draw = cv2.rectangle(draw, (0, 0), (200, 200), (255, 255, 255), 3)
draw = cv2.putText(draw, 'Hello CV2', (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)
```

```python
cv2.imshow('Image', draw)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
