```python
import torch
import torch.nn as nn
```

```python
# создаём слой
layer = nn.Linear(5, 5)
```

```python
# начальные веса
layer.weight.data
```

```python
# равномерное распределение
nn.init.uniform_(layer.weight, a=0, b=3)
```

```python
# номральное распределение
nn.init.normal_(layer.weight, mean=0, std=1)
```

```python
# можно константой, но очень нежелательно в большей части случаев
# можно, например, для инициализации смещения
nn.init.constant_(layer.bias, val=2)
# или аналоги для 0 и 1
# nn.init.zeros_(layer.bias)
# nn.init.ones_(layer.bias)
```

```python
# инициализация Xavier
nn.init.xavier_normal_(layer.weight, gain=1)
```
