```python
import torch
import torch.nn as nn # модуль для нейросетей
```

```python
# выводим 5 переменных для 4 экземпляров
prediction = torch.randn(4, 5)
```

```python
# реальные значения
label = torch.randn(4, 5)
```

```python
# определяем функцию потерб
# reduction - то, как устреднять ошибку по батчам
mse = torch.nn.MSELoss(reduction='none')
```

```python
# считаем MSE
loss = mse(prediction, label)
```

```python
loss
```

```python
mse = torch.nn.MSELoss(reduction='mean')
loss = mse(prediction, label)
loss
```

```python
# математическая формула
((prediction - label) ** 2).mean()
```

Бинарная кросс энтропия

```python
prediction
```

```python
# сделали массив label нулями, а потом заплнили его ослучайными числами из [0, 1]
label = torch.zeros(4, 5).random_(0, 1+1)
```

```python
label
```

```python
# сигмоида нужна, чтобы привести prediction к 0, 1
sigmoid = nn.Sigmoid()
```

```python
# определяем функию потерь
bce = nn.BCELoss(reduction='mean')
```

```python
bce(sigmoid(prediction), label)
```

```python
# попробуем bce с сигмоидой
bces = torch.nn.BCEWithLogitsLoss(reduction='mean')
```

```python
bces(prediction, label)
```

Реализация на python

```python
import numpy as np
```

```python
# конвертируем
x = prediction.numpy()
y = label.numpy()
```

```python
# приводим к [0, 1]
def my_sigmoid(x):
    return 1 / (1 + np.exp(-x))
x = my_sigmoid(x)
x
```

```python
# считаем
loss_values = []
for i in range(len(y)):
    batch_loss = []
    for j in range(len(y[0])):
        batch_loss.append((-np.log(x[i][j])) if y[i][j] == 1 else (-np.log(1 - x[i][j])))
    loss_values.append(batch_loss)
np.mean(loss_values)
```
