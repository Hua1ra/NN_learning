```python
from PIL import Image
import os
from random import randint
import numpy as np
import matplotlib.pyplot as plt
```

```python
# загружаем картинки
img_folder = "source/images/"
imgs = os.listdir(img_folder)
batch_x = [Image.open(img_folder + p).resize((224, 224)) for p in imgs]
```

```python
# нормаилзация
def normalize_image(x):
    x = x / np.max(x)
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    return (x - mean) / std
```

```python
# денормализация
def denormalize_image(x):
    mean = np.array([0.485, 0.456, 0.406])
    std = np.array([0.229, 0.224, 0.225])
    x = std * x + mean
    x = np.clip(x, 0, 1)
    return x
```

```python
# два изображения
lam = 0.7
batch_size = len(batch_x)
current_image = normalize_image(batch_x[0])
random_index = randint(1, len(batch_x) - 1)
random_image_from_batch = normalize_image(np.array(batch_x[random_index]))
```

```python
plt.imshow(denormalize_image(current_image))
plt.axis("off")
```

```python
plt.imshow(denormalize_image(random_image_from_batch))
plt.axis("off")
```

```python
mixed_image = lam * current_image  + (1 - lam) * random_image_from_batch
```

```python
plt.imshow(denormalize_image(mixed_image))
```
