```python
images = []
for i, item in enumerate(bins):
    if isinstance(item, bytes):
        img = Image.open(io.BytesIO(item)).convert('RGB')
        images.append(img)
    # elif isinstance(item, np.ndarray):
    #     img = Image.fromarray(item).convert('RGB')
    #     images.append(img)
```

```python
print(len(images))
print(len(is_same_list))
print(sum(is_same_list))
for i, item in enumerate(images):
    plt.axis('off')
    plt.imshow(item)
    plt.show()
    if i % 2 == 1:
        print(is_same_list[i // 2])
    if (i + 1) % 10 == 0:
        break
```
