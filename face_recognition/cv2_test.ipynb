```python
import cv2 # type: ignore
from ultralytics import YOLO # type: ignore
```

```python
model = YOLO('models/yolov8n-face.pt')
```

```python
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    exit()
while True:
    ret, frame = cap.read()
    if not ret:
        break
    results = model(frame, verbose=False)
    left_eye = tuple(map(int, results[0].keypoints.xy[0][0].tolist()))
    right_eye = tuple(map(int, results[0].keypoints.xy[0][1].tolist()))
    nose = tuple(map(int, results[0].keypoints.xy[0][2].tolist()))
    x1, y1, x2, y2 = map(int, results[0].boxes[0].xyxy[0])
    cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
    cv2.drawMarker(frame, left_eye, (0, 255, 0), 3)
    cv2.drawMarker(frame, right_eye, (0, 255, 0), 3)
    cv2.drawMarker(frame, nose, (0, 255, 0), 1)
    cv2.imshow('Face Recognition Camera Test', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```
