```python
import psycopg2 # type: ignore
import numpy as np # type: ignore
```

```python
host = 'localhost'
database = 'face_recognition'
username = 'postgres'
password = 'postgres'
port = 5432
```

```python
conn = psycopg2.connect(host=host,
                        database=database,
                        user=username,
                        password=password,
                        port=port)
cur = conn.cursor()
```

```python
first_name = 'Alexey'
last_name = 'Korolyuk'
surname = 'Leonidovich'
birthdate = '2006-10-11'
cur.execute(f'''
    INSERT INTO person(first_name, last_name, surname, birthdate)
    VALUES
    ('{first_name}', '{last_name}', '{surname}', '{birthdate}');
''')
conn.commit()
```

```python
cur.execute('''
    SELECT *
    FROM person;
''')
rows = cur.fetchall()
for row in rows:
    print(row)
```

```python
cur.execute('''
    TRUNCATE TABLE person CASCADE;
''')
conn.commit()
```
